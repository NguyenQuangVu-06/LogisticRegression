**Let observe how LogLoss changes over 5 iterations**

```python
import math
import numpy as np
import matplotlib.pyplot as plt
X = [1, 9, 10, 11, 14, 21, 24, 27, 30]
y = [0, 0, 1, 1, 0, 0, 1, 1 ,0]
w , b = 0 , 0
learning_rate = 0.01
epochs = 100000
losses = []
for e in range(epochs):
    lst_z = []
    for i in range(len(X)):
        z = w * X[i] + b
        lst_z.append(z)
    lst_sigmoid = []
    for i in range(len(X)):
        sigmoid = 1/(1+math.exp(-lst_z[i]))
        lst_sigmoid.append(sigmoid)
    lst_log_loss = []
    for i in range(len(X)):
        log_loss =  y[i]*math.log(lst_sigmoid[i])+(1-y[i])*math.log(1-lst_sigmoid[i])
        lst_log_loss.append(log_loss)
    Log_Loss = np.sum(lst_log_loss)
    Log_loss = -1/len(X)*Log_Loss
    losses.append(Log_loss)
    g_w , g_b = 0 , 0
    for i in range(len(X)):
        g_w += X[i]*(lst_sigmoid[i]-y[i])
        g_b += lst_sigmoid[i] - y[i]
    g_w = g_w/len(X)
    g_b = g_b/len(X)
    w = w - learning_rate*g_w
    b = b - learning_rate*g_b
fig , ax = plt.subplots(nrows=1,ncols=2,figsize=(10,10))
ax[0].plot(range(epochs),losses,color="red")
ax[0].set(xlabel='Epochs',ylabel='Loss',title='Loss Function')
ax[1].scatter(X,y,color="red",label="Data")
ax[1].plot(X,lst_sigmoid,color="blue",label="Sigmoid")
ax[1].set(xlabel="X",ylabel="Y",title="Sigmoid Function")
ax[1].legend(loc="lower right")
plt.show()
```

**We can see the LogLoss decrease after each epoch**
