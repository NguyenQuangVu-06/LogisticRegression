```python
import math
import numpy as np
import matplotlib.pyplot as plt
X = [ #Hour , Sleep
    [1,8],
    [2,7],
    [3,6],
    [4,7],
    [5,8],
    [6,6],
    [7,7],
    [8,6]
]
y = [0,0,0,0,1,1,1,1]  #Pass or Not Pass
w1,w2,b=0,0,0
epochs = 1000
losses = []
for e in range(epochs):
    lst = []
    for i in range(len(X)):
        z = w1*X[i][0] + w2*X[i][1] + b
        lst.append(z)
    p = []
    for i in range(len(X)):
        sigmoid = 1/(1+math.exp(-lst[i]))
        p.append(sigmoid)
    log_loss = 0
    for i in range(len(X)):
        log_loss += y[i]*math.log(p[i]) + (1-y[i])*math.log(1-p[i])
    log_loss = -log_loss/len(X)
    losses.append(log_loss)
    g_w1 , g_w2 , g_b = 0, 0 ,0
    for i in range(len(X)):
        g_w1 += X[i][0]*(p[i]-y[i])
        g_w2 += X[i][1]*(p[i]-y[i])
        g_b += p[i] - y[i]
    g_w1 = g_w1/len(X)
    g_w2 = g_w2/len(X)
    g_b = g_b/len(X)
    w1 = w1 - 0.01*g_w1
    w2 = w2 - 0.01*g_w2
    b = b - 0.01*g_b
plt.plot(range(epochs),losses,color="yellow")
plt.xlabel('Epochs')
plt.ylabel('Log_Loss')
plt.title('Training')
plt.show()
X = np.array(X)
plt.scatter(X[:,0],y,color="red")
plt.plot(X[:,0],p,color="blue",label="Hour")
plt.xlabel("Hour")
plt.ylabel("Pass or Not Pass")
plt.legend()
plt.show()
```
