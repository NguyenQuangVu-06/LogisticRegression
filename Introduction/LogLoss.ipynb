**To see whether model is very good , we rely on log-loss**

**L=−[ylog(p)+(1−y)log(1−p)] when has only 1 element**

```python
import math
#Assume y = 1 , sigmoid = p = 0.8
y , p = 1 , 0.8
log_loss = -(y*math.log(p)+(1-y)*math.log(1-p))
print(log_loss)
```

```python
#Case 2 : y , p = 1 , 0.2
y1,p1=1,0.2
log_loss1 = -(y1*math.log(p1)+(1-y1)*math.log(1-p1))
print(log_loss1)
```

**We can see when p is increases then logloss is decrease , this is great**
