**We will try calculated sigmoid when z = [1,2,3,4]**

```python
import math
z = [1,2,3,4]
for i in z:
    sigmoid = 1/(1+math.exp(-i))
    print(sigmoid)
```

**We will try with z = 1000 , 10000 to see if sigmoid exceeds 1**

```python
import math
z = [1000,10000,100000]
for i in z:
    sigmoid = 1/(1+math.exp(-i))
    print(sigmoid)
```
