**We have many probabilities such as 0.12 0.31 0.49 0.50 0.51 0.78 0.9**

**We will use threshold = ? to divide probabilities to a specific class ?**


**Example : threshold = 0.2 , we will put on X in class 0**

```python
probability = [0.12,0.31,0.49,0.50,0.51,0.78,0.9]
threshold = 0.5
#0.12 < 0.5 => class 0
#0.31 < 0.5 => class 0
# .....
# 0.51 > 0.5 => class 1
# 0.78 > 0.5 => class 1
# Probabilities is calculated by sigmoid function
```

 **A threshold = 0.5 commonly used because it at a midpoint , making classification to easier**

```python
import math
X = [1, 2, 3, 4, 5]
# Assume w, b = 1 , -3
w , b = 1 , -3
for i in X:
    sigmoid = 1/(1+math.exp(-(w*i+b)))
    print(sigmoid)
```

**Assume threshold = 0.5 then has 2 class 0 and 3 class 1**

**Note : When Sigmoid = Threshold then we select priority class 1**
