*We will find w , b is most suitable for multi loops*

```python
import math
X = [1, 9, 10, 11, 14, 21, 24, 27, 30]
y = [0, 0, 1, 1, 0, 0, 1, 1 ,0]
w , b = 0 , 0
epochs = 5
learning_rate = 0.01
for e in range(epochs):
    p = []
    g_w , g_b = 0 , 0
    log_loss = 0
    for i in range(len(X)):
        sigmoid = 1/(1+math.exp(-(w*X[i]+b)))
        p.append(sigmoid)
    for i in range(len(X)):
        log_loss += y[i]*math.log(p[i]) + (1-y[i])*math.log(1-p[i])
    log_loss = -1/len(X)*log_loss
    for i in range(len(X)):
        g_w += X[i]*(p[i]-y[i])
        g_b += p[i] - y[i]
    g_w = g_w/len(X)
    g_b = g_b/len(X)
    w = w - learning_rate*g_w
    b = b - learning_rate*g_b
    print(w,b,log_loss)
```

*We can use after each iteration , log_loss decrease , this is great*
