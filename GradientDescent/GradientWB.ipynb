```python
import math
X = [1, 2, 3]
y = [0, 0, 1]
#Assem both w , b are 0
w = 0
b = 0
z = []
for i in range(len(X)):
    a = w*X[i] + b
    z.append(a)
lst_sigmoid = []
for i in z:
    sigmoid = 1/(1+math.exp(-i))
    lst_sigmoid.append(sigmoid)
g_w = 0
g_b = 0
for i in range(len(lst_sigmoid)):
    g_w += (lst_sigmoid[i]-y[i])*X[i]
    g_b += lst_sigmoid[i] - y[i]
g_b = g_b/len(X)
g_w = g_w/len(X)
print(g_w,g_b)
```
